Just a collection of miscallaneous figures we'll be using within our report.


```python
import matplotlib.pyplot as plt
import seaborn as sns
```

### Plot accuracy with 20 epochs

```python
# Plot the average accuracy over number of epochs
avg_acc = [0.286, 0.429, 0.429, 0.429, 0.429, 0.500, 0.643, 0.643, 0.643, 0.643, 0.643, 0.643, 0.643, 0.643, 0.643, 0.643, 0.643, 0.714, 0.714, 0.714]
epochs = list(range(1, len(avg_acc)+1))
sns.set_style("whitegrid")
sns.set_palette("deep")
ax = sns.scatterplot(x=epochs, y=avg_acc, color="gray")
sns.lineplot(x=epochs, y=avg_acc)

ax.set_title(f"Training Accuracy Over {len(avg_acc)} Epochs")
ax.set_xlabel("Number of Epochs")
ax.set_ylabel("Average Accuracy")
ax.set_xticks(list(range(0, len(avg_acc)+1, 5)))
```
